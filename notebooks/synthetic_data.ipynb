```python
tools_df=pd.read_sql("SELECT * FROM tools",engine)
machines_df=pd.read_sql("SELECT * FROM machines",engine)
operators_df=pd.read_sql("SELECT * FROM operators",engine)
workorders_df=pd.read_sql("SELECT * FROM work_orders",engine)
toolpositions_df=pd.read_sql("SELECT * FROM tool_positions",engine)
```

## 2. Master Data Generation

```python
import random
```

```python
random.choice(tools_df["insert_shape"])
```

```python
rows=[]
for i in range(100):
    operation_id= f"{i:04d}"
    work_order_id= random.choice(workorders_df["work_order_id"].tolist())
    machine_id= random.choice(machines_df["machine_id"].tolist())
    operator_id= random.choice(operators_df["operator_id"].tolist())
    parts= random.randint(20,100)
    cutting_time= random.uniform(1,10)
    rpm=random.randint(800,3500)

    rows.append([operation_id,work_order_id,machine_id,operator_id,parts,cutting_time,rpm])

```

```python
rows
```

```python
operations_df=pd.DataFrame(rows)
```

```python
operations_df.head()
```

```python
operations_df = pd.DataFrame(rows,columns=["operation_id","work_order_id","machine_id","operator_id","parts","cutting_time","rpm"])
```

## 3. Operations Data Generation

```python
operation_tool_rows= []
```

```python
for operation_id in operations_df["operation_id"]:
    num_tools = random.randint(1, 3)
    selected_tools = random.sample(tools_df["tool_id"].tolist(), num_tools)

    for tool_id in selected_tools:
        operation_tool_rows.append([operation_id, tool_id])
```

## 4. Operation-Tool Relationships

```python
operation_tools_df= pd.DataFrame(operation_tool_rows,columns=["operation_id","tool_id"])
```

```python
operation_tools_df.shape
```

```python
#operation_tools_df.to_sql("operation_tools",engine,if_exists="append",index=False)
```

```python
tool_positions_df=pd.read_sql("SELECT * FROM tool_positions",engine)
```

```python
tool_positions_df.head()
```

```python
operation_tools_df = pd.read_sql("SELECT * FROM operation_tools",engine)
```

```python
operation_tools_df.head()
```

```python
tool_positions_df.head()
```

```python
operation_positions_df=operation_tools_df.merge(
    tool_positions_df,
    on="tool_id",
    how="inner"
)
```

```python
operation_positions_df.head()
```

```python
operation_positions_df = operation_positions_df.groupby(
    ["operation_id", "tool_id"]
).sample(n=1)
```

```python
operation_positions_df.head(10)
```

```python
operations_df=pd.read_sql("SELECT * FROM operations",engine)
```

```python
wear_base_df=operation_positions_df.merge(
    operations_df,
    on="operation_id",
    how="left"
)
```

```python
wear_base_df.head(10)
```

## 5. Wear Measurement Generation

```python
wear_values = []

for _, row in wear_base_df.iterrows():
    wear = (
        row["parts"] * 0.001
        + row["cutting_time"] * 0.01
        + random.uniform(0, 0.03)
    )
    
    wear_values.append(wear)
```

```python
wear_base_df["wear_mm"]=wear_values
```

```python
wear_base_df.head()
```

```python
conditions=[]

for wear in wear_base_df["wear_mm"]:
    if wear<0.10:
        condition="Normal"
    elif wear<0.20:
        condition="Slight Whitening"
    else:
        condition="Visible Wear"

conditions.append(condition)
```

```python
conditions = []

for wear in wear_base_df["wear_mm"]:
    if wear < 0.10:
        condition = "Normal"
    elif wear < 0.20:
        condition = "Slight Whitening"
    else:
        condition = "Visible Wear"

    conditions.append(condition)

print(len(conditions))
```

```python
wear_base_df["visual_condition"] = conditions
```

```python
wear_base_df.head(10)
```

```python
wear_base_df.columns
```

```python
wear_measurements_df = wear_base_df[
    ["position_id", "operation_id", "wear_mm", "visual_condition"]
].copy()
```

```python
wear_measurements_df.head()
```

## 6. Date Generation

```python
from datetime import datetime, timedelta

start_date = datetime(2026, 1, 1)

dates = []

for i in range(len(wear_measurements_df)):
    dates.append(start_date + timedelta(days=i))
```

```python
wear_measurements_df["mesaurement_date"] = dates
```
